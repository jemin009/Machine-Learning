```python
df = pd.read_csv(r"C:\Users\jemin\Downloads\Datamites\trainRF.csv")

df
```

```python
#1 Display phones where ram is greater than 3000 and battery_power is greater than 1500.

show = df.loc[(df['ram'] > 3000) & (df['battery_power'] > 1500)]

show
```

```python
#2 Calculate the average value of the ram column. 

avg = df['ram'].mean()

avg
```

```python
#3 Count how many phones belong to each price_range category. 

count = df['price_range'].value_counts()

count
```

```python
#4 Identify the maximum RAM value and display the row of the phone with that value. 

max = df.loc[df['ram'] == df['ram'].max()]

max
```

```python
#5 Display all unique values in price_range and count how many unique categories exist. 

uniq = df['price_range'].unique()

count = df['price_range'].nunique()

print(uniq)
print(count)
```

```python
#6 Count missing values in the ram, battery_power, and int_memory columns. 

miss = df[['ram','battery_power','int_memory']].isnull().sum()

print(miss)
```

```python
#7 Count the duplicate rows in the dataset. Do not remove them. 

dup = df.duplicated().sum()

dup
```

```python
#8 Calculate the average ram for each price_range category. 

avg = df.groupby('price_range')['ram'].mean()

avg
```

```python
#9 Create X using ram, battery_power, and int_memory. Create y using price_range. Print both shapes. 

x = df[['ram','battery_power','int_memory']]
y = df['price_range']

print(x.shape)
print(y.shape)
```

```python
#10 Filter phones where price_range is 3. Display the number of matching phones and their average RAM. 

show = df.loc[df['price_range'] == 3]

count = show.shape[0]
avg_ram = show['ram'].mean()

print("Number of phones:", count)
print("Average RAM:", avg_ram)
```
