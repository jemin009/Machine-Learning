```python
#6 Split `X` and `y` into training and testing sets using `train_test_split`. Use `test_size=0.2` and `random_state=42`.

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    x, y,
    test_size = 0.2,
    random_state = 42
)
```

```python
#9 Display the first five rows of the scaled training features. Use the variable names from Question 8.

X_train_scaled[:5]
```

```python
#10 Use `LogisticRegression` to train a classification model on the training data from Question 8. Predict `price_range` for `X_test` and display the first 10 predictions.

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

print(y_pred[:10])
```

```python
#6 
"""
A data scientist wants to select a model for predicting phone price categories. Use 5-fold cross-validation to compare: 

 LogisticRegression(max_iter=1000) — use scaled features. 
 KNeighborsClassifier(n_neighbors=5) — use scaled features. 
 RandomForestClassifier(n_esmators=100, random_state=42) — use unscaled features. 

Print the mean accuracy for each model. Use a Pipeline for the scaled models to avoid data leakage between folds. 
"""

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# Logistic Regression with scaling
model_log = make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000))

cv_log = cross_val_score(model_log, X_train, y_train, cv=5, scoring='accuracy')

# KNN with scaling
model_knn = make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5))

cv_knn = cross_val_score(model_knn, X_train, y_train, cv=5, scoring='accuracy')

# Random Forest without scaling
model_rf = RandomForestClassifier(n_estimators=100, random_state=42)

cv_rf = cross_val_score(model_rf, X_train, y_train, cv=5, scoring='accuracy')

# Print mean accuracies
print(f"Logistic Regression: {cv_log.mean() * 100:.2f}%")
print(f"KNN:                {cv_knn.mean() * 100:.2f}%")
print(f"Random Forest:      {cv_rf.mean() * 100:.2f}%")
```
