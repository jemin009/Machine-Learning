```python
#6 Split `X` and `y` into training and testing sets using `train_test_split`. Use `test_size=0.2` and `random_state=42`.

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    x, y,
    test_size = 0.2,
    random_state = 42
)
```

```python
#9 Display the first five rows of the scaled training features. Use the variable names from Question 8.

X_train_scaled[:5]
```

```python
#10 Use `LogisticRegression` to train a classification model on the training data from Question 8. Predict `price_range` for `X_test` and display the first 10 predictions.

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

print(y_pred[:10])
```

```python
#1 
"""
Compare Three Models 

Logistic Regression, KNN aur Random Forest train karo. Har model ki test accuracy calculate karke compare karo. 

Concept: accuracy_score(), predict()
"""

from sklearn.linear_model import LogisticRegression 
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

model_lr = make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000))
model_lr.fit(X_train, y_train)
pred_lr = model_lr.predict(X_test)
accu_lr = accuracy_score(y_test, pred_lr)

model_knn = KNeighborsClassifier(n_neighbors=5)
model_knn.fit(X_train, y_train)
pred_knn = model_knn.predict(X_test)
accu_knn = accuracy_score(y_test, pred_knn)

model_rf = RandomForestClassifier(n_estimators=100, random_state=42)
model_rf.fit(X_train, y_train)
pred_rf = model_rf.predict(X_test)
accu_rf = accuracy_score(y_test, pred_rf)

print(f"Accuracy of Logistic model is {accu_lr * 100:.2f}%")
print(f"Accuracy of knn model is {accu_knn * 100:.2f}%")
print(f"Accuracy of RandomForest model is {accu_rf * 100:.2f}%")
```

```python
#2 
"""
Compare Mean Cross-Validation Accuracy 

Teeno models par 5-fold cross-validation run karo aur unki mean accuracy compare karo. 

Concept: cross_val_score(), mean()
"""

from sklearn.model_selection import cross_val_score

cv_lr = cross_val_score(model_lr, X_train, y_train, cv=5, scoring='accuracy')
cv_knn = cross_val_score(model_knn, X_train, y_train, cv=5, scoring='accuracy')
cv_rf = cross_val_score(model_rf, X_train, y_train, cv=5, scoring='accuracy')

print(f"Mean accuracy of Logistic model is {cv_lr.mean() * 100:.2f}%")
print(f"Mean accuracy of knn model is {cv_knn.mean() * 100:.2f}%")
print(f"Mean accuracy of RandomForest model is {cv_rf.mean() * 100:.2f}%")
```

```python
#3 
"""
Compare Model Consistency 

Har model ke five fold scores ka highest–lowest difference calculate karo. Identify karo kaunsa model sabse consistent hai. 

Concept: max(), min(), difference
"""

print(f"Difference of Logistic model is {abs(cv_lr.min() - cv_lr.max()) * 100:.2f}%")
print(f"Difference of knn model is {abs(cv_knn.min() - cv_knn.max()) * 100:.2f}%")
print(f"Difference of RandomForest model is {abs(cv_rf.min() - cv_rf.max()) * 100:.2f}%")
```

```python
#4 
"""
Create a Model Comparison Table 

Ek Pandas DataFrame banao jisme har model ki test accuracy, mean CV accuracy aur fold difference ho. 

Concept: pd.DataFrame() 
"""

import pandas as pd

comparison = pd.DataFrame({
    'Model': ['Logistic Regression', 'KNN', 'Random Forest'],

    'Test Accuracy': [
        accu_lr * 100,
        accu_knn * 100,
        accu_rf * 100
    ],

    'Mean CV Accuracy': [
        cv_lr.mean() * 100,
        cv_knn.mean() * 100,
        cv_rf.mean() * 100
    ],

    'Fold Difference': [
        (cv_lr.max() - cv_lr.min()) * 100,
        (cv_knn.max() - cv_knn.min()) * 100,
        (cv_rf.max() - cv_rf.min()) * 100
    ]
})

print(comparison.round(2))
```

```python
#5
"""
Choose the Best Model by Accuracy 

Comparison table se identify karo ki kis model ki mean CV accuracy sabse zyada hai. Us model ka naam aur score print karo. 

Concept: idxmax()
"""

best_model = comparison.loc[comparison['Mean CV Accuracy'].idxmax()]

print(f"Best model: {best_model['Model']}")
print(f"Mean CV Accuracy: {best_model['Mean CV Accuracy']:.2f}%")
```

```python
#6 
"""
Compare Precision, Recall and F1-Score 

Teeno models ke macro precision, macro recall aur macro F1-score ko cross-validation se calculate karke compare karo. 

Concept: cross_validate()
"""

from sklearn.model_selection import cross_validate

cv_lr = cross_validate(model_lr,X_train,y_train,cv=5,scoring=['precision_macro', 'recall_macro', 'f1_macro'])
cv_knn = cross_validate(model_knn,X_train,y_train,cv=5,scoring=['precision_macro', 'recall_macro', 'f1_macro'])
cv_rf = cross_validate(model_rf,X_train,y_train,cv=5,scoring=['precision_macro', 'recall_macro', 'f1_macro'])

print(cv_lr)
print(cv_knn)
print(cv_rf)
```

```python
#7 
"""
Business Requirement: Consistency 

Ek retailer chahta hai ki model ki performance stable rahe. Aisa model identify karo jiski mean CV accuracy achhi ho aur fold difference relatively low ho.

Concept: Accuracy vs consistency
"""

from sklearn.metrics import accuracy_score

print(comparison.round(2))

# Here you can see that Logistic Model has Best mean cv accuracy with lowest fold difference so Business can use Logistic model
```

```python
#8 
"""
Business Requirement: Avoid Wrong Predictions

Maan lo retailer ke liye kisi particular price_range category ko miss karna costly hai. 
Models ke per-class recall ko compare karo aur identify karo ki kaunsa model us category ko sabse achha identify karta hai. 

Concept: recall_score(), classification_report()
"""

from sklearn.metrics import recall_score, classification_report

pred_lr = model_lr.predict(X_test)
pred_knn = model_knn.predict(X_test)
pred_rf = model_rf.predict(X_test)

print("Recall_Lr:", recall_score(y_test, pred_lr, average='macro'))
print("Recall_Knn:", recall_score(y_test, pred_knn, average='macro'))
print("Recall_Rf:", recall_score(y_test, pred_rf, average='macro'))

print(classification_report(y_test, pred_lr))
print(classification_report(y_test, pred_knn))
print(classification_report(y_test, pred_rf))
```

```python
#9 
"""
Select a Model Using Multiple Criteria 

Aisa model choose karo jo mean CV accuracy, test accuracy aur consistency ke beech achha balance deta ho.
Apne actual results ke basis par decision explain karo. 

Concept: Evidence-based model select on
"""

best_model = comparison.loc[comparison['Mean CV Accuracy'].idxmax()]

print("Selected Model:", best_model['Model'])
print("Test Accuracy:", best_model['Test Accuracy'])
print("Mean CV Accuracy:", best_model['Mean CV Accuracy'].round(2))
print("Fold Difference:", best_model['Fold Difference'].round(2))
```

```python
#10
"""
Final Business Recommendation 

Retailer ke liye ek short recommendation likho: selected model ka naam, uski accuracy, CV consistency, important metric aur use choose karne ka reason. 

Concept: Model select on + business interpretation
"""

import pandas as pd

best = pd.DataFrame({
    'Model' : [best_model['Model']],
    'Accuracy' : [best_model['Test Accuracy']],
    'Consistency' : [best_model['Mean CV Accuracy']],
    'Fold Difference' : [best_model['Fold Difference']]
})

print(best)

print(
    "\nRecommendation: Logistic Regression is the preferred model "
    "\nbecause it has the highest test accuracy (81.75%), "
    "\nthe highest mean CV accuracy (82.38%), and the lowest "
    "\nfold difference (3.44 percentage points). "
    "\nIt offers the best balance of predictive performance "
    "\nand consistency among the three models tested."
)

```

```python
#6
"""
Duplicate-Record Leakage 

Duplicate rows identify karo aur samjho ki similar records train aur test dono mein aane se evaluation kaise misleading ho sakta hai. 

Concept: duplicated(), data splitting
"""

import numpy as np

unique_train = np.unique(X_train, axis=0)
train_duplicates = len(X_train) - len(unique_train)

print("Train duplicate rows:", train_duplicates)


unique_test = np.unique(X_test, axis=0)
test_duplicates = len(X_test) - len(unique_test)

print("Test duplicate rows:", test_duplicates)


train_rows = set(map(tuple, X_train))
test_rows = set(map(tuple, X_test))

common_records = train_rows.intersection(test_rows)

print("Common train-test records:", len(common_records))
```

```python
#7
"""
Incorrect vs Correct Evaluation 

Leaky preprocessing aur correct pipeline ke evaluation results compare karo. 

Concept: Reliable model evaluation
"""

import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score

# A: Incorrect approach — scaling before splitting
scaler_A = StandardScaler()
X_scaled = scaler_A.fit_transform(x)

X_train_A, X_test_A, y_train_A, y_test_A = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

model_A = LogisticRegression(max_iter=1000)
model_A.fit(X_train_A, y_train_A)

pred_A = model_A.predict(X_test_A)
accuracy_A = accuracy_score(y_test_A, pred_A)


# B: Correct approach — split first, then use a pipeline
model_B = make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000))

model_B.fit(X_train, y_train)
pred_B = model_B.predict(X_test)

accuracy_B = accuracy_score(y_test, pred_B)


# Compare results
print(f"Incorrect approach accuracy: {accuracy_A * 100:.2f}%")
print(f"Correct approach accuracy:   {accuracy_B * 100:.2f}%")
print(f"Difference: {abs(accuracy_A - accuracy_B) * 100:.2f} percentage points")
```
