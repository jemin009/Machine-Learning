```python
df = pd.read_csv(r"C:\Users\jemin\Downloads\Datamites\trainRF.csv")

df
```

```python
#1 Create `X` using `ram`, `battery_power`, and `int_memory`. Create `y` using `price_range`. Display the first five rows of both. 

x = df[['ram','battery_power','int_memory']]
y = df['price_range']

print(x.head())
print(y.head())
```

```python
#2 Check the number of missing values in every column of `X`. 

miss = x.isnull().sum()

miss
```

```python
#3 Create a new DataFrame named `df_clean` by filling missing values in `ram` with the median RAM. Do not modify the original `df`. 

df_clean = df.copy()

df_clean['ram'] = df_clean['ram'].fillna(df_clean['ram'].median())

df_clean
```

```python
#4 Create a new DataFrame named `df_unique` by removing duplicate rows. Display the number of rows before and after removal.

print(f"Rows before removal: {df.shape[0]}")

df_unique = df.drop_duplicates()

print(f"Rows after removal: {df_unique.shape[0]}")
```

```python
#5 Create `X` using `ram` and `battery_power`, and create `y` using `price_range`. Display the shape of `X`. 

x = df[['ram','battery_power']]
y = df['price_range']

x.shape
```

```python
#6 Split `X` and `y` into training and testing sets using `train_test_split`. Use `test_size=0.2` and `random_state=42`.

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    x, y,
    test_size = 0.2,
    random_state = 42
)
```

```python
#7 Print the shapes of `X_train`, `X_test`, `y_train`, and `y_test` from Question 6. 

print(f"shape of X_train : {X_train.shape}")
print(f"shape of X_test : {X_test.shape}")
print(f"shape of y_test : {y_test.shape}")
print(f"shape of y_train : {y_train.shape}")
```

```python
#8 Use `StandardScaler` to scale `X_train` and `X_test`. Fit the scaler on training data only, then transform both datasets. 

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

```

```python
#9 Display the first five rows of the scaled training features. Use the variable names from Question 8.

X_train_scaled[:5]
```

```python
#10 Use `LogisticRegression` to train a classification model on the training data from Question 8. Predict `price_range` for `X_test` and display the first 10 predictions.

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

print(y_pred[:10])
```

# ML5


```python
#1 Use `accuracy_score` to compare `y_test` and `y_pred`. Print the accuracy. 

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
```

```python
#2 Convert the accuracy score into a percentage and print it with the `%` symbol. 

convert = accuracy * 100

print(f"{convert}%")
```

```python
#3 Use `confusion_matrix` to compare actual labels `y_test` with predictions `y_pred`. Display the matrix.

from sklearn.metrics import confusion_matrix

compare = confusion_matrix(y_test, y_pred)

print(compare)
```

```python
#4 Use `classification_report` to display precision, recall, F1-score and support for each price category. 

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)

print(report)
```

```python
#5 Create a comparison between `y_test` and `y_pred`. Count how many predictions are correct and how many are incorrect. 

correct = (y_test == y_pred).sum()
incorrect = (y_test != y_pred).sum()

print("Correct Predictions:", correct)
print("Incorrect Predictions:", incorrect)
```

```python
#6 Find the test rows where `y_pred` does not match `y_test`. Display the first 5 incorrect predictions alongside their actual labels.

incorrect = y_test != y_pred

result = X_test[incorrect].copy()

result["Actual"] = y_test[incorrect]
result["Predicted"] = y_pred[incorrect]

print(result.head(5))
```

```python
#7 Calculate the accuracy separately for each actual `price_range` category (0, 1, 2 and 3).

from sklearn.metrics import accuracy_score

for category in [0, 1, 2, 3]:
    mask = y_test == category

    accuracy = accuracy_score(y_test[mask], y_pred[mask])

    print(f"Price Range {category} Accuracy: {accuracy:.2%}")
```

```python
#8 Train a `KNeighborsClassifier` using `X_train_scaled` and `y_train`. Predict labels for `X_test_scaled` and store them in `y_pred_knn`.

from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(X_train_scaled, y_train)

y_pred_knn = knn.predict(X_test_scaled)

print(y_pred_knn[:10])
```

```python
#9 Calculate and print the accuracy of Logistic Regression (`y_pred`) and KNN (`y_pred_knn`). Identify which has the higher test accuracy. 

from sklearn.metrics import accuracy_score

log_accuracy = accuracy_score(y_test, y_pred) * 100
knn_accuracy = accuracy_score(y_test, y_pred_knn) * 100

print(f"Accuracy of Logistic Regression : {log_accuracy}%")
print(f"Accuracy of KNN : {knn_accuracy}%")

if log_accuracy > knn_accuracy:
    print("Logistic Regression has higher test accuracy.")
elif knn_accuracy > log_accuracy:
    print("KNN has higher test accuracy.")
else:
    print("Both models have equal test accuracy.")
```
